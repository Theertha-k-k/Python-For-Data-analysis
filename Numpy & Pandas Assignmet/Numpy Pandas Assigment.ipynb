```python
import numpy as np
```

```python
# 1 .Create a 1D numpy array
temperatures_w1 = np.array([22.5,25.3,20.8,23.4,26.1,24.8,21.9])
print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# 2. Findout shape,datatype,number of elements and dimension
print("Shape:",temperatures_w1.shape)
print("Data Type:",temperatures_w1.dtype)
print("Number of Elements:",temperatures_w1.size)
print("Array Dimenson:",temperatures_w1.ndim)
```

```python
# 3. Array operations
# Convert the temperatures_w1 from celcius to Farenheit
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

```python
# Find maximum ,minimum,and mean temperatures

maximum_temp = np.max(temperatures_w1)
print("Maximum Temperature:",maximum_temp)

minimum_temp = np.min(temperatures_w1)
print("Minimun Temperature:",minimum_temp)

mean_temp = np.mean(temperatures_w1)
print("Mean Temperature:",mean_temp)
```

```python
# 4. Array slicing and indexing
# Extract the temperatures for the first three days of the week. 
first_three_days = temperatures_w1[0:3]
print("First Three Days:")
print(first_three_days)

# Extract the temperatures for the weekend (last two days).
weekend_temp = temperatures_w1[-2:]
print("Weekend Temperatures:")
print(weekend_temp)

#Extract the temperatures for the middle three days of the 
middle_three_days = temperatures_w1[2:5]
print("Middle Three Days:")
print(middle_three_days)
```

```python
# 5 Create a 2D numpy array named temperatures where each row represents temperatures for a different week
import numpy as np
temperatures = np.array([
    [22.5,25.3,20.8,23.4,26.1,24.8,21.9],
    [19.2,22.5,21.3,24.0,23.5,22.8,20.1]
])

print("temperatures:")
print(temperatures)
```

```python
# 6. Inspect and slice the 2D array
# Inspect shape,data type,total number of elements
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
print("Number of Dimensions:", temperatures.ndim)
```

```python
# Extract temperatures for each week and weekends
week1 = temperatures[0,:]
print("Week 1 Temperatures:",week1)
```

```python
# Extract week 2 temperatures
week2 = temperatures[1, :]
print("Week 2 Temperatures:",week2)
```

```python
# Extract weekend temperatures for both weeks
weekend = temperatures[:,-2:]
print("Weekend Temperatures for Both Weeks:",weekend)
```

```python
# PANDAS Series
# 1. Creating pandas series
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
# 2.  Indexing and slicing
# to findout  mark of the 1st rank student
print("Mark of 1 st Rank Student:",marks.iloc[0])
```

```python
# to retrieve the marks of top 3 ranks
print("Marks of Top Ranks:")
print(marks.loc[["Rank1","Rank2","Rank3"]])
```

```python
# to retrieve the mark of the 3 rd rank student
print("Mark of 3 rd Rank Student:",marks.iloc[2])
```

```python
# to retrieve the ranks where the marks are greater than 90
print("Students scoring Above 90:")
print(marks[marks >90])
```

```python
# 3. Manipulating pandas series
# Modify the mark of the student with the 1 st rak to 100
marks.loc["Rank1"] = 100
print("Updated Marks:")
print(marks)
```

```python
# Remove the entry corresponding to the last rank
marks = marks.drop("Rank5")
print("Marks After Removing Last Rank:")
print(marks)
```

```python
# Compute the CGPA by dividing each mark by 10
cgpa = marks / 10
print("CGPA of Students:")
print(cgpa)
```

```python
# PANDAS DATAFRAME:
#1. Creating a DataFrame
# TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110 
# ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics' 
# Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North' 
# Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105,
                      106, 107, 108, 109, 110],

    'ProductCategory': ['Electronics', 'Clothing', 'Electronics',
                        'Furniture', 'Clothing', 'Electronics',
                        'Furniture', 'Clothing', 'Furniture',
                        'Electronics'],

    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],

    'Amount': [200, 150, 300, 450, 200,
               250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print("Transaction DataFrame:")
print(transactions)
```

```python
# 2. Data Exploration
# A. Display the DataFrame
print(transactions)
```

```python
# B. Dsplay the first 5 rows 
print("First 5 Rows:")
print(transactions.head())
```

```python
# C. Display the last 5 rows
print("Last 5 Rows:")
print(transactions.tail())
```

```python
# D . Display the shape of the DataFrame
print("Shape:", transactions.shape)
```

```python
#E . Display column names
print("Column Names:")
print(transactions.columns)
```

```python
# F. Display data types
print("Data Types:")
print(transactions.dtypes)
```

```python
# G . Display only product category and amount
print("Product Category and Amount:")
print(transactions[['ProductCategory', 'Amount']])
```

```python
# Retrieve the last 3 columns of the DataFrame using 
print("Last 3 Columns:")
print(transactions.iloc[:, -3:])
```

```python
# Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200
filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]
print("Filtered Transactions:")
print(filtered_data)
```

```python
# Find value counts for ProductCategory
print("Product Category Counts:")
print(transactions['ProductCategory'].value_counts())
```

```python
# Find unique values in the Region column
print("Unique Regions:")
print(transactions['Region'].unique())
```

```python
# Group by Region and calculate the mean Amount
print("Mean Transaction Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. Manipulating the DataFrame
# Modify Amount for TransactionID 102 to 165
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print("Updated Transaction:")
print(transactions)
```

```python
# Add a new column Discount by calculating 10% of Amount
transactions['Discount'] = transactions['Amount'] * 0.10
print("DataFrame After Adding Discount:")
print(transactions)
```

```python
# Remove the row with TransactionID 109
transactions = transactions.drop(
    transactions[transactions['TransactionID'] == 109].index
)

print("DataFrame After Removing Transaction 109:")
print(transactions)
```

```python
# Delete the Discount column from the DataFrame
transactions = transactions.drop(columns=['Discount'])

print("Final DataFrame:")
print(transactions)
```
